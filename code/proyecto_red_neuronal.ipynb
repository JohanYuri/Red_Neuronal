# Importamos Pandas y leemos el dataset

```python
import pandas as pd

# Definimos el archivo csv como una variable df y lo leemos con la función read_csv de pandas
df = pd.read_csv('../Data/dataset_heart.csv')

# Mostramos las primeras 5 filas
df.head(5)
```

# Inspeccionamos los datos

```python
print("*** Información del Dataset ***")
print(df.info())
print("\n*** Conteo de Valores Nulos ***")
print(df.isnull().sum())
```

# Ajustamos la variable objetivo (Target)

```python
# Transformamos: 1 (Sin enfermedad) -> 0, y 2 (Con enfermedad) -> 1
df['heart disease'] = df['heart disease'].replace({1: 0, 2: 1})

# Verificamos que el cambio se hizo correctamente
print("\n*** Distribución de la variable objetivo (0 y 1) ***")
print(df['heart disease'].value_counts())
```
